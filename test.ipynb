```python
import os
import cv2
import numpy as np
from tensorflow.keras.models import load_model

# -------------------------------
# Paths
# -------------------------------
model_path = r"C:\Users\namit\Documents\emo_proj\emotion_detection_model.h5"
img_path = r"C:\Users\namit\Documents\emo_proj\test\disgust\Training_659019.jpg"

# -------------------------------
# Load Model
# -------------------------------
print("🔄 Loading model...")
model = load_model(model_path)
print("✅ Model loaded successfully!\n")

# -------------------------------
# Emotion Labels
# -------------------------------
emotion_labels = ['angry', 'disgust', 'fear', 'happy', 'neutral', 'sad', 'surprise']

# -------------------------------
# Read & Preprocess Image
# -------------------------------
if not os.path.exists(img_path):
    raise FileNotFoundError(f"❌ Image not found at: {img_path}")

print(f"🖼️ Testing image: {img_path}")

# Read image in grayscale
img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
if img is None:
    raise ValueError("⚠️ Could not read the image. Check the file format or path.")

# Resize and normalize
img = cv2.resize(img, (48, 48))
img = img / 255.0
img = np.expand_dims(img, axis=(0, -1))  # shape (1, 48, 48, 1)

# -------------------------------
# Predict
# -------------------------------
prediction = model.predict(img, verbose=0)
predicted_class = np.argmax(prediction)
predicted_emotion = emotion_labels[predicted_class]

# -------------------------------
# Output
# -------------------------------
print(f"\n✅ Predicted Emotion: {predicted_emotion}")
```

```python
import os
import cv2
import numpy as np
from tensorflow.keras.models import load_model

# -------------------------------
# Paths
# -------------------------------
model_path = "emotion_detection_model.h5"
test_dir = "test"  # Folder inside emo_proj

# -------------------------------
# Load the trained model
# -------------------------------
print("🔄 Loading model...")
model = load_model(model_path)
print("✅ Model loaded successfully!\n")

# Emotion labels (adjust based on your dataset)
emotion_labels = ["angry", "disgust", "fear", "happy"]

# -------------------------------
# Function to predict emotion from one image
# -------------------------------
def predict_emotion(image_path):
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    if img is None:
        print(f"⚠️ Skipping invalid image: {image_path}")
        return

    img = cv2.resize(img, (48, 48))
    img = img / 255.0
    img = np.expand_dims(img, axis=(0, -1))

    prediction = model.predict(img)
    predicted_class = emotion_labels[np.argmax(prediction)]
    confidence = np.max(prediction) * 100

    print(f"🖼️ Image: {os.path.basename(image_path)} → 😃 Predicted: {predicted_class} ({confidence:.2f}%)")
```
