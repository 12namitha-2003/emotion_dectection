```python
import os
import cv2
from tqdm import tqdm

# Define dataset paths
data_dir = "data/train"
cleaned_dir = "data_cleaned/train"

# Create output folder if not exists
os.makedirs(cleaned_dir, exist_ok=True)

# Loop through emotion folders
for emotion in os.listdir(data_dir):
    emotion_path = os.path.join(data_dir, emotion)
    cleaned_emotion_path = os.path.join(cleaned_dir, emotion)
    os.makedirs(cleaned_emotion_path, exist_ok=True)

    print(f"🧹 Cleaning images in '{emotion}'...")

    for img_file in tqdm(os.listdir(emotion_path)):
        img_path = os.path.join(emotion_path, img_file)

        try:
            img = cv2.imread(img_path)
            if img is None:
                continue

            gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            resized = cv2.resize(gray, (48, 48))
            save_path = os.path.join(cleaned_emotion_path, img_file)
            cv2.imwrite(save_path, resized)

        except Exception as e:
            print(f"Error with {img_file}: {e}")

print("\n✅ Data cleaning complete! Cleaned images saved in:", cleaned_dir)
```
