```python
import os
import shutil
from sklearn.model_selection import train_test_split

# Define paths
cleaned_dir = "data_cleaned/train"
output_train = "data_cleaned/train"
output_test = "data_cleaned/test"

# Create test directory
os.makedirs(output_test, exist_ok=True)

# Loop through each emotion folder
for emotion in os.listdir(cleaned_dir):
    emotion_path = os.path.join(cleaned_dir, emotion)
    images = os.listdir(emotion_path)

    # Split 80% train, 20% test
    train_imgs, test_imgs = train_test_split(images, test_size=0.2, random_state=42)

    # Create emotion subfolders for train/test
    train_emotion_dir = os.path.join(output_train, emotion)
    test_emotion_dir = os.path.join(output_test, emotion)
    os.makedirs(test_emotion_dir, exist_ok=True)

    # Copy test images
    for img in test_imgs:
        src = os.path.join(emotion_path, img)
        dst = os.path.join(test_emotion_dir, img)
        shutil.copy(src, dst)

print("✅ Data successfully split into 'train' and 'test' folders!")
```
